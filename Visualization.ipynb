```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Generate values for the parameter t
t = np.linspace(-2 * np.pi, 2 * np.pi, 500)

# Define the parametric equations for x, y, and z (a helix)
x = np.sin(t)
y = np.cos(t)
z = t

# Parameters for the tube
radius = 0.2  # Radius of the tube
num_points_circle = 20 # Number of points to define the circle cross-section

# Create points for the tube surface
theta = np.linspace(0, 2 * np.pi, num_points_circle) # Angle for the circle

# Initialize arrays for the tube surface coordinates
x_tube = np.zeros((len(t), num_points_circle))
y_tube = np.zeros((len(t), num_points_circle))
z_tube = np.zeros((len(t), num_points_circle))

for i in range(len(t)):
    # Normal vector for the cross-section (e.g., perpendicular to the tangent of the helix)
    # For a simple helix, we can rotate the circle around the helix's center axis
    x_tube[i, :] = x[i] + radius * np.cos(theta)
    y_tube[i, :] = y[i] + radius * np.sin(theta)
    z_tube[i, :] = z[i] * np.ones(num_points_circle)

# Create the 3D figure
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Plot the tube surface
surf = ax.plot_surface(x_tube, y_tube, z_tube, cmap='viridis', edgecolor='none', alpha=0.9)

# Add labels and title
ax.set_title('3D Parametric Helix (Tube-like)')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')

# Adjust the viewing angle for better perspective
ax.view_init(elev=20, azim=45)

# Make the background panes transparent
ax.xaxis.pane.fill = False
ax.yaxis.pane.fill = False
ax.zaxis.pane.fill = False

# Add a color bar
fig.colorbar(surf, shrink=0.5, aspect=5)

plt.show()
```

###  Surface Plot with Projections

This option displays the surface with contour projections on the base planes, which can help visualize the function's level sets.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Assuming X, Y, Z are already defined from the previous cell
# If not, you might need to re-run the previous cell or define them here.
X = np.arange(-5, 5, 0.25)
Y = np.arange(-5, 5, 0.25)
X, Y = np.meshgrid(X, Y)
Z = np.sin(np.sqrt(X**2 + Y**2))

# Create the 3D figure
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
surf = ax.plot_surface(X, Y, Z, cmap='coolwarm', alpha=0.8)

# Add contour projections onto the planes
cset = ax.contourf(X, Y, Z, zdir='z', offset=ax.get_zlim()[0], cmap='coolwarm', alpha=0.5)
cset = ax.contourf(X, Y, Z, zdir='x', offset=ax.get_xlim()[0], cmap='coolwarm', alpha=0.5)
cset = ax.contourf(X, Y, Z, zdir='y', offset=ax.get_ylim()[1], cmap='coolwarm', alpha=0.5)

# Add labels and title
ax.set_title('3D Surface Plot with Projections')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')

# Add a color bar
fig.colorbar(surf, shrink=0.5, aspect=5)

# Show the plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Parameters for the Mobius Strip
u = np.linspace(0, 2 * np.pi, 100)  # Angle around the strip
v = np.linspace(-0.5, 0.5, 50)  # Width across the strip
u, v = np.meshgrid(u, v)

# Parametric equations for a Mobius strip
# R is the radius of the 'circle' around which the strip is formed
# r is the half-width of the strip itself (controlled by v)
R = 2
x_mobius = (R + v * np.cos(u / 2)) * np.cos(u)
y_mobius = (R + v * np.cos(u / 2)) * np.sin(u)
z_mobius = v * np.sin(u / 2)

# Create the 3D figure
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Plot the Mobius surface
surf = ax.plot_surface(x_mobius, y_mobius, z_mobius, cmap='coolwarm', edgecolor='none')

# Add labels and title
ax.set_title('3D Parametric Mobius Strip')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')

# Set aspect ratio for better visualization (optional, but good for 3D shapes)
ax.set_box_aspect([np.ptp(x_mobius), np.ptp(y_mobius), np.ptp(z_mobius)])

# Add a color bar
fig.colorbar(surf, shrink=0.5, aspect=5)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Parameters for the Klein Bottle
v_klein = np.linspace(0, 2 * np.pi, 100)  # Main 'loop' parameter
u_klein = np.linspace(0, 2 * np.pi, 50)  # Cross-section parameter
v_klein, u_klein = np.meshgrid(v_klein, u_klein)

# Parametric equations for a Klein Bottle
# These equations are quite complex and define its self-intersecting shape

# First part of the Klein Bottle (the 'handle')
# This portion is more challenging to visualize directly as a single function

# A simpler, more common visualization uses a "figure-8" immersion
# However, the full mathematical Klein bottle is harder to plot as a single surface
# We'll use a commonly adapted parametric form for visualization:

r = 4 * (1 - np.cos(u_klein) / 2)

x_klein = 6 * np.cos(u_klein) * (1 + np.sin(u_klein)) + r * np.cos(v_klein + np.pi)
y_klein = 16 * np.sin(u_klein) + r * np.sin(v_klein + np.pi)

z_klein = r * np.cos(v_klein)

# Adjust for the self-intersection visually (common approach for plotting)
# This is a piecewise definition to avoid 'folding' issues in basic plotting
x_klein[u_klein > np.pi] = 6 * np.cos(u_klein[u_klein > np.pi]) * (1 + np.sin(u_klein[u_klein > np.pi])) + r[u_klein > np.pi] * np.cos(v_klein[u_klein > np.pi])
y_klein[u_klein > np.pi] = 16 * np.sin(u_klein[u_klein > np.pi])
z_klein[u_klein > np.pi] = r[u_klein > np.pi] * np.sin(v_klein[u_klein > np.pi] + np.pi)

# Create the 3D figure
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Plot the Klein Bottle surface
surf = ax.plot_surface(x_klein, y_klein, z_klein, cmap='viridis', edgecolor='none', alpha=0.8)

# Add labels and title
ax.set_title('3D Parametric Klein Bottle')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')

# Set aspect ratio for better visualization
ax.set_box_aspect([np.ptp(x_klein), np.ptp(y_klein), np.ptp(z_klein)])

# Add a color bar
fig.colorbar(surf, shrink=0.5, aspect=5)

# Adjust viewing angle for better perspective
ax.view_init(elev=20, azim=60) # elev: elevation angle, azim: azimuthal angle

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Parameters for the Torus
R_torus = 3  # Major radius
r_torus = 1  # Minor radius

u = np.linspace(0, 2 * np.pi, 100)  # Angle around the major circle
v = np.linspace(0, 2 * np.pi, 50)   # Angle around the minor circle
u, v = np.meshgrid(u, v)

# Parametric equations for a Torus
x_torus = (R_torus + r_torus * np.cos(v)) * np.cos(u)
y_torus = (R_torus + r_torus * np.cos(v)) * np.sin(u)
z_torus = r_torus * np.sin(v)

# Create the 3D figure
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Plot the Torus surface
surf = ax.plot_surface(x_torus, y_torus, z_torus, cmap='plasma', edgecolor='none', alpha=0.9)

# Add labels and title
ax.set_title('3D Parametric Torus')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')

# Set aspect ratio for better visualization
ax.set_box_aspect([np.ptp(x_torus), np.ptp(y_torus), np.ptp(z_torus)])

# Add a color bar as a legend for Z-values
fig.colorbar(surf, shrink=0.5, aspect=5, label='Z-value (Minor Circle Height)')

# Adjust viewing angle for better perspective
ax.view_init(elev=20, azim=30)

plt.show()
```
