```python
# 15. Function to check if a number is prime
def functions_is_prime(n):
  # This function checks if a given number 'n' is a prime number.
  print("Anya Jain 0827CI231023")
  if n < 2:
    return False
  for i in range(2, int(n**0.5) + 1):
    if n % i == 0:
      return False
  return True

print(f"Is 7 prime? {functions_is_prime(7)}")
print(f"Is 10 prime? {functions_is_prime(10)}")
```

```python
# 16. Function to reverse a string
def functions_reverse_string(s):
  # This function reverses the input string 's' using slicing.
  print("Anya Jain 0827CI231023")
  return s[::-1]

print(f"Reversed 'hello': {functions_reverse_string('hello')}")
```

```python
# 17. Function to find the maximum in a list
def functions_find_max(numbers):
  # This function finds and returns the maximum value from a list of numbers.
  print("Anya Jain 0827CI231023")
  if not numbers:
    return None
  return max(numbers)

print(f"Max of [1, 5, 2, 8]: {functions_find_max([1, 5, 2, 8])}")
```

```python
# 18. Function to count vowels in a string
def functions_count_vowels(s):
  # This function counts the number of vowels (a, e, i, o, u) in the input string 's'.
  print("Anya Jain 0827CI231023")
  vowels = "aeiouAEIOU"
  count = 0
  for char in s:
    if char in vowels:
      count += 1
  return count

print(f"Vowels in 'Programming': {functions_count_vowels('Programming')}")
```

```python
# 19. Function to calculate average of a list of numbers
def functions_calculate_average(numbers):
  # This function calculates the average of numbers in a given list.
  print("Anya Jain 0827CI231023")
  if not numbers:
    return 0
  return sum(numbers) / len(numbers)

print(f"Average of [10, 20, 30]: {functions_calculate_average([10, 20, 30])}")
```

```python
# 20. Function to check if a string is a palindrome
def functions_is_palindrome(s):
  # This function checks if a string is a palindrome (reads the same forwards and backward).
  print("Anya Jain 0827CI231023")
  cleaned_s = "".join(filter(str.isalnum, s)).lower()
  return cleaned_s == cleaned_s[::-1]

print(f"Is 'Madam' a palindrome? {functions_is_palindrome('Madam')}")
print(f"Is 'hello' a palindrome? {functions_is_palindrome('hello')}")
```

```python
# 21. Function to get unique elements from a list
def functions_get_unique_elements(lst):
  # This function returns a list containing only the unique elements from the input list 'lst'.
  print("Anya Jain 0827CI231023")
  return list(set(lst))

print(f"Unique elements of [1, 2, 2, 3, 4, 4]: {functions_get_unique_elements([1, 2, 2, 3, 4, 4])}")
```

```python
# 22. Function to merge two dictionaries
def functions_merge_dicts(d1, d2):
  # This function merges two dictionaries 'd1' and 'd2'.
  print("Anya Jain 0827CI231023")
  merged = d1.copy()
  merged.update(d2)
  return merged

dict1 = {'a': 1, 'b': 2}
dict2 = {'b': 3, 'c': 4}
print(f"Merged dicts: {functions_merge_dicts(dict1, dict2)}")
```

```python
# 23. Function to calculate factorial using a loop
def functions_factorial_loop(n):
  # This function calculates the factorial of a non-negative integer 'n' using a loop.
  print("Anya Jain 0827CI231023")
  if n < 0:
    return "Factorial not defined for negative numbers"
  if n == 0:
    return 1
  res = 1
  for i in range(1, n + 1):
    res *= i
  return res

print(f"Factorial of 5 (loop): {functions_factorial_loop(5)}")
```

```python
# 24. Function to check if a year is a leap year
def functions_is_leap_year(year):
  # This function determines if a given 'year' is a leap year.
  print("Anya Jain 0827CI231023")
  return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

print(f"Is 2024 a leap year? {functions_is_leap_year(2024)}")
print(f"Is 1900 a leap year? {functions_is_leap_year(1900)}")
```

```python
# 25. Function to convert Celsius to Fahrenheit
def functions_celsius_to_fahrenheit(celsius):
  # This function converts a temperature from Celsius to Fahrenheit.
  print("Anya Jain 0827CI231023")
  return (celsius * 9/5) + 32

print(f"25 Celsius to Fahrenheit: {functions_celsius_to_fahrenheit(25)}")
```

```python
# 26. Function to convert Fahrenheit to Celsius
def functions_fahrenheit_to_celsius(fahrenheit):
  # This function converts a temperature from Fahrenheit to Celsius.
  print("Anya Jain 0827CI231023")
  return (fahrenheit - 32) * 5/9

print(f"77 Fahrenheit to Celsius: {functions_fahrenheit_to_celsius(77)}")
```

```python
# 27. Function to calculate the sum of numbers in a range
def functions_sum_range(start, end):
  # This function calculates the sum of all integers within a specified range (inclusive).
  print("Anya Jain 0827CI231023")
  return sum(range(start, end + 1))

print(f"Sum of numbers from 1 to 5: {functions_sum_range(1, 5)}")
```

```python
# 28. Function to count word occurrences in a string
def functions_word_count(text):
  # This function counts the occurrences of each word in a given text string.
  print("Anya Jain 0827CI231023")
  words = text.lower().split()
  counts = {}
  for word in words:
    counts[word] = counts.get(word, 0) + 1
  return counts

print(f"Word counts for 'hello world hello': {functions_word_count('hello world hello')}")
```

```python
# 29. Function to capitalize the first letter of each word
def functions_capitalize_words(s):
  # This function capitalizes the first letter of each word in a string.
  print("Anya Jain 0827CI231023")
  return ' '.join(word.capitalize() for word in s.split())

print(f"Capitalized 'this is a test': {functions_capitalize_words('this is a test')}")
```

```python
# 30. Function to check if all elements in a list are unique
def functions_are_all_unique(lst):
  # This function checks if all elements in a list are unique.
  print("Anya Jain 0827CI231023")
  return len(lst) == len(set(lst))

print(f"Are all unique in [1, 2, 3]? {functions_are_all_unique([1, 2, 3])}")
print(f"Are all unique in [1, 2, 2]? {functions_are_all_unique([1, 2, 2])}")
```

```python
# 31. Function to flatten a list of lists
def functions_flatten_list(list_of_lists):
  # This function flattens a list of lists into a single list.
  print("Anya Jain 0827CI231023")
  return [item for sublist in list_of_lists for item in sublist]

print(f"Flattened [[1, 2], [3, 4]]: {functions_flatten_list([[1, 2], [3, 4]])}")
```

```python
# 32. Function to generate Fibonacci sequence up to n terms
def functions_fibonacci_sequence(n):
  # This function generates the Fibonacci sequence up to 'n' terms.
  print("Anya Jain 0827CI231023")
  if n <= 0:
    return []
  elif n == 1:
    return [0]
  else:
    list_fib = [0, 1]
    while len(list_fib) < n:
      next_fib = list_fib[-1] + list_fib[-2]
      list_fib.append(next_fib)
    return list_fib

print(f"Fibonacci sequence up to 7 terms: {functions_fibonacci_sequence(7)}")
```

```python
# 33. Function to power of a number
def functions_power(base, exponent):
  # This function calculates the power of a 'base' number raised to an 'exponent'.
  print("Anya Jain 0827CI231023")
  return base ** exponent

print(f"2 to the power of 3: {functions_power(2, 3)}")
```

#### 5. Function with Arbitrary Positional Arguments (`*args`)

```python
# 39. Function with Arbitrary Positional Arguments (*args)
print("Anya Jain 0827CI231023")
def multiply_all(*numbers):
  # This function calculates the product of an arbitrary number of arguments.

  product = 1
  for num in numbers:
    product *= num
  return product

print(f"Product of 2, 3, 4 is: {multiply_all(2, 3, 4)}")
print(f"Product of 5, 10 is: {multiply_all(5, 10)}")
```

#### 6. Function with Arbitrary Keyword Arguments (`**kwargs`)

```python
# 40. Function with Arbitrary Keyword Arguments (**kwargs)
def print_profile(**data):
  # This function prints user profile data passed as arbitrary keyword arguments.
  print("Anya Jain 0827CI231023")
  for key, value in data.items():
    print(f"{key.replace('_', ' ').title()}: {value}")

print_profile(name="David", city="New York", occupation="Engineer")
```

#### 7. Function that Returns Multiple Values

```python
# 41. Function that Returns Multiple Values
def calculate_stats(numbers):
  # This function calculates and returns the total and average of a list of numbers.
  print("Anya Jain 0827CI231023")
  total = sum(numbers)
  average = total / len(numbers)
  return total, average

scores = [85, 90, 78, 92]
sum_scores, avg_scores = calculate_stats(scores)
print(f"Total scores: {sum_scores}, Average scores: {avg_scores:.2f}")
```

#### 8. Lambda Function (Anonymous Function)

```python
# 42. Lambda Function (Anonymous Function)
# This cell demonstrates lambda functions for squaring a number and sorting a list of tuples.
print("Anya Jain 0827CI231023")
square = lambda x: x * x
print(f"Square of 7 is: {square(7)}")

# Using lambda with sorted()
points = [(1, 2), (0, 5), (3, 1)]
sorted_points = sorted(points, key=lambda p: p[1])
print(f"Sorted points by y-coordinate: {sorted_points}")
```

#### 9. Function with Conditional Logic

```python
# 43. Function with Conditional Logic
def check_even_odd(number):
  # This function checks if a given number is even or odd.

  if number % 2 == 0:
    return f"{number} is Even"
  else:
    return f"{number} is Odd"

print(check_even_odd(4))
print(check_even_odd(7))
print("Anya Jain 0827CI231023")
```

#### 10. Recursive Function (Factorial Example)

```python
# 44. Recursive Function (Factorial Example)
print("Anya Jain 0827CI231023")
def factorial(n):
  # This function calculates the factorial of a number recursively.

  if n == 1:
    return 1
  else:
    return n * factorial(n-1)

print(f"Factorial of 5 is: {factorial(5)}")
```
